```python
# Modulo:  tema9_proyDiccionarios.py

# Mi Base de datos

menu = {1:"Agregar Producto",
        2:"Retirar Producto",
        3:"Modificar Producto",
        4:"Listar Productos",
        5:"Aumentar Stock",
        6:"Reducir Stock",
        7:"Ventas",
        8:"Comprobante",
        9:"Salir del Sistema"
        }

clientes = {47878988:"Salas Rivera Juan",
            41223255:"Manco Lopez Ana",
            54878988:"Verástegui Ramos Juliana",
            45845888:"Sanchez Castro Filemon"
            }

direccion = {47878988:"Av. Pacasmayo 123 Callao",
             41223255:"Av. Perú 1234 SMP",
             54878988:"Jr. Tacna 234 Ate Vitarte",
             45845888:"Psj. Olaya 2343 Lince"}

empleados = {1000:"Lopez Carpio Gian Marco",
             1001:"Sanchez Gómez Patricia",
             1002:"Ramos Espejo Joel",
             1003:"Robles Alcantara Elizabeth"}


producto = {1:"Arroz",2:"Azucar",3:"Fideos",4:"Harina",5:"Aceite", 6:"Sal",
            7:"Huevos", 8:"Galletas", 9:"Gelatina",10:"Frejol",11:"Canela"}

precio = {1:4.0, 2:5.0, 3:1.5, 4:8.0, 5:11.0, 6:1.0, 7:7.50, 8:1.5,
          9:2.5, 10:7.50, 11:1.5}

stock = {1:50, 2:50, 3:50, 4:25, 5:50, 6: 35, 7: 150.0, 8: 25, 9: 30,
         10:20, 11: 30}

detalle = {} #<---- diccionario destinado para el comprobante.



# Modulos def

def listaProducto():
    print()
    print("-------------------------------------------------")
    print("Lista de Productos en Almacén")
    print("-------------------------------------------------")
    for i in producto:
        print(i, producto[i], precio[i], stock[i])
    print("-------------------------------------------------")
    print("Fin del Listado......")
    salida = input("Presione Enter para salir..")
    print()

def agregarProducto():
    print()
    print("-------------------------------------------------")
    print("Agregar Productos en Almacén")
    print("-------------------------------------------------")
    clave = max(producto) + 1
    producto[clave] = input("Ingrese Producto: ")
    precio[clave] = float(input("Ingrese Precio: "))
    stock[clave] = int(input("Ingrese Stock: "))
    print("Producto Agregado......")
    print()
    salida = input("Presione Enter para salir...")

def retirarProducto():
    print()
    print("-------------------------------------------------")
    print("Retirar Productos de Almacén")
    print("-------------------------------------------------")
    for i in producto:
        print(i, producto[i], precio[i], stock[i])
    print("-------------------------------------------------")
    clave = int(input("Seleccione el producto: "))
    if (clave in producto):
        del producto[clave]
        del precio[clave]
        del stock[clave]
        print("Producto retirado del almacen....")
        print()
    else:
        print("El producto seleccionado no existe..")
        print()

    salida = input("Presione Enter pasa salir....")
    print()

def modificarProducto():
    print()
    print("-------------------------------------------------")
    print("Modificar Productos de Almacén")
    print("-------------------------------------------------")
    for i in producto:
        print(i, producto[i], precio[i], stock[i])
    print("-------------------------------------------------")
    clave = int(input("Seleccione el producto: "))
    if (clave in producto):
        producto[clave] = input("Modificar Nombre: ")
        precio[clave] = float(input("Modificar Precio: "))
        stock[clave] = int(input("Modificar Stock: "))
        print("Producto Modificado en almacen....")
        print()
    else:
        print("El producto seleccionado no existe..")
        print()

    salida = input("Presione Enter pasa salir....")
    print()

def aumentarStock():
    print()
    print("-------------------------------------------------")
    print("Aumentar Stock de Productos de Almacén")
    print("-------------------------------------------------")
    for i in producto:
        print(i, producto[i], precio[i], stock[i])
    print("-------------------------------------------------")
    clave = int(input("Seleccione el producto: "))
    if (clave in producto):
        print("Producto        : ", producto[clave])
        print("Precio          : ", precio[clave])
        print("Stock   ------->: ", stock[clave])
        print("-------------------------------------------------")
        cantidad = int(input("Ingrese Cantidad: "))
        stock[clave] += cantidad
        print("Stock de Producto Modificado en almacen....")
        print()
    else:
        print("El producto seleccionado no existe..")
        print()

    salida = input("Presione Enter pasa salir....")
    print()

def reducirStock():
    print()
    print("-------------------------------------------------")
    print("Reducir Stock de Productos de Almacén")
    print("-------------------------------------------------")
    for i in producto:
        print(i, producto[i], precio[i], stock[i])
    print("-------------------------------------------------")
    clave = int(input("Seleccione el producto: "))
    if (clave in producto):
        print("Producto        : ", producto[clave])
        print("Precio          : ", precio[clave])
        print("Stock   ------->: ", stock[clave])
        print("-------------------------------------------------")
        cantidad = int(input("Ingrese Cantidad: "))
        if (cantidad <= stock[clave]):
            stock[clave] -= cantidad
            print("Stock de Producto Modificado en almacen....")
            print()
        else:
            print()
            print("Error: cantidad sobrepasa lo establecido..")


    else:
        print("El producto seleccionado no existe..")
        print()

    salida = input("Presione Enter pasa salir....")
    print()


def ventas():
    print()
    print("-------------------------------------------------")
    print("VENTAS:")
    print("-------------------------------------------------")
    for i in producto:
        e0 = 4 - len(str(i))
        e1 = 25 - len(producto[i])
        e2 = 5 - len(str(precio[i]))
        print(i,(e0*" "), producto[i],(e1*" "), precio[i],(e2*" "),stock[i])
    print("-------------------------------------------------")
    clave = int(input("Seleccione el producto a vender: "))
    cantidad = int(input("Ingrese Cantidad: "))
    item = len(detalle) + 1
    detalle[item] = [producto[clave], precio[clave], cantidad, precio[clave]*cantidad]
    print(detalle)
    print()

def comprobante():
    print()
    print("-------------------------------------------------")
    print("COMPROBANTE DE PAGO:")
    print("-------------------------------------------------")
    dni = int(input("Ingrese Dni del Cliente: "))
    ce =  int(input("Ingrese clave de Empleado: "))
    if dni in clientes:
        nombreC = clientes[dni]
        domicilio = direccion[dni]
        if ce in empleados:
            nombreE = empleados[ce]
            print("-------------------------------------------------")
            print("Nro Comprobante      : 000000001")
            print("Dni                  : ", dni)
            print("Nombre Cliente       : ", nombreC)
            print("Dirección Cliente    : ", domicilio)
            print("Clave del Empleado   : ", ce)
            print("Nombre del Empleado  : ", nombreE)
            print("-------------------------------------------------")
            for i in detalle:
                e1
                print(i, detalle[i][0], detalle[i][1], detalle[i][2], detalle[i][3])
            print("-------------------------------------------------")
            print()







# Programa Principal

while(True):
    print("-------------------------------------------------")
    for i in range(1, len(menu)+1):
        print(i, menu[i])
    print("-------------------------------------------------")
    opcion = int(input("Selecciones una opción: "))
    if (opcion == 1):
        agregarProducto()

    elif (opcion == 2):
        retirarProducto()

    elif (opcion == 3):
        modificarProducto()

    elif (opcion == 4):
        listaProducto()

    elif (opcion == 5):
        aumentarStock()

    elif (opcion == 6):
        reducirStock()

    elif (opcion == 7):
        ventas()

    elif (opcion == 8):
        comprobante()

    elif (opcion == 9):
        break

```
