```python
# Modulo: tema9_excepciones1.py

while (True):
    try:
        a = int(input("Ingrese Primer valor : "))
        break
    except:
        print()
        print("Error: , el valor es incorrecto, vuelva a intentar")
        print()

while (True):
    try:
        b = int(input("Ingrese Segundo valor: "))
        break
    except:
        print()
        print("Error: , el valor es incorrecto, vuelva a intentar")
        print()
try:
    c = a / b
    print("---------------------------------------")
    print("Prueba de división")
    print("---------------------------------------")
    print("Valor de A        : ", a)
    print("Valor de B        : ", b)
    print("División          : ", c)
except:
    print()
    print("Error: , No se puede dividir por cero...")
    print()
```
